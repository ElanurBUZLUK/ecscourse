```python
def istatistik_hesapla(sayilar):
    """
    Bir sayı listesinin temel istatistiklerini hesaplar.
    """

    toplam = sum(sayilar)
    en_kucuk_val = min(sayilar)
    en_buyuk_val = max(sayilar)
    ortalama = toplam / len(sayilar)

    return toplam, en_kucuk_val, en_buyuk_val, ortalama
```

```python
notlar=[50,60,80,80]
istatistik_hesapla(notlar)[0]
```

```python
toplam_val, min_val, max_val, ort_val = istatistik_hesapla(notlar)

print(
    f"Toplam: {toplam_val}, "
    f"En küçük: {min_val}, "
    f"En büyük: {max_val}, "
    f"Ortalama: {ort_val}"
)
```

```python
def faktoriyel(n):
    sonuc = 1

    for i in range(1, n + 1):
        sonuc *= i

    return sonuc
```

```python
def asalkontrol(n):
    if (n) < 2:
        return False
    elif n == 2:
        return True
    for i in range(2, int((n/2))+2):
        if n % i == 0:
            return False
    return True
```

```python
asalkontrol(2)
```

```python
def maskele(ad, soyad, kimlik_no):
    ad_msk = ad[:2] + "****"
    soyad_msk = soyad[:2] + "****"
    kimlik_no_msk = kimlik_no[:3] + "********"
    return ad_msk, soyad_msk, kimlik_no_msk
```

```python
maskele("Elanur", "Buzluk", "12345678901")
```
